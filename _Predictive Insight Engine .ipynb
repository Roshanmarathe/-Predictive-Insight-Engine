## ⚙ Part G: Gradient Descent Optimization

### Q21.Gradient Descent

Gradient Descent is an optimization technique used to reduce the error of a model.

It repeatedly changes the model parameters in the direction that reduces the loss.

- Batch GD: Uses all data at once.
- SGD: Uses one data point at a time.
- Mini-Batch GD: Uses a small group of data points.

### Q22. Implement Batch Gradient Descent from Scratch

```python
from sklearn.preprocessing import StandardScaler
import time

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
y_train_arr = np.asarray(y_train)

def Batch_GD(X, y, lr=0.01, epochs=100):
    n, f = X.shape
    W = np.zeros(f)
    b = 0
    losses = []

    for _ in range(epochs):
        pred = X @ W + b
        error = pred - y
        losses.append(np.mean(error**2))
        W -= lr * (2/n) * X.T @ error
        b -= lr * (2/n) * np.sum(error)

    return W, b, losses

start = time.time()
w_batch, b_batch, loss_batch = Batch_GD(X_train_scaled, y_train_arr)

print("Final MSE:", loss_batch[-1])
print("Training Time:", time.time()-start)
```

### Q23. Implement Stochastic Gradient Descent (SGD)

```python
def Stochastic_GD(X, y, lr=0.001, epochs=20):
    n, f = X.shape
    W = np.zeros(f)
    b = 0
    losses = []

    for _ in range(epochs):
        for i in np.random.permutation(n):
            error = X[i] @ W + b - y[i]
            W -= lr * 2 * error * X[i]
            b -= lr * 2 * error
        losses.append(np.mean((X @ W + b - y)**2))

    return W, b, losses

start = time.time()
w_sgd, b_sgd, loss_sgd = Stochastic_GD(X_train_scaled, y_train_arr)

print("Final MSE:", loss_sgd[-1])
print("Training Time:", time.time()-start)
```

### Q24. Implement Mini-Batch Gradient Descent

```python
def Mini_Batch_GD(X, y, lr=0.01, epochs=30, batch_size=32):
    n, f = X.shape
    W = np.zeros(f)
    b = 0
    losses = []

    for _ in range(epochs):
        idx = np.random.permutation(n)

        for i in range(0, n, batch_size):
            batch = idx[i:i+batch_size]
            error = X[batch] @ W + b - y[batch]
            W -= lr * (2/len(batch)) * X[batch].T @ error
            b -= lr * (2/len(batch)) * np.sum(error)

        losses.append(np.mean((X @ W + b - y)**2))

    return W, b, losses

start = time.time()
w_mb, b_mb, loss_mb = Mini_Batch_GD(X_train_scaled, y_train_arr)

print("Final MSE:", loss_mb[-1])
print("Training Time:", time.time()-start)
```

```python
import matplotlib.pyplot as plt

plt.plot(loss_batch, label="Batch GD")
plt.plot(loss_sgd, label="SGD")
plt.plot(loss_mb, label="Mini-Batch GD")

plt.xlabel("Epoch")
plt.ylabel("Training MSE")
plt.title("Batch vs SGD vs Mini-Batch")
plt.yscale("log")
plt.legend()
plt.show()
```

### Observation

Batch GD has smooth convergence because it uses all training data.

SGD makes frequent updates and usually shows more fluctuations.

Mini-Batch GD provides a balance between stable convergence and faster updates.

Training time can vary depending on the number of epochs and batch size.

### Q27. Explain How Model Complexity Affects Prediction Error

### Model Complexity

As model complexity increases, the model can learn more complex patterns.

- Low complexity → high bias → underfitting.
- Moderate complexity → better balance between bias and variance.
- Very high complexity → high variance → overfitting.

Therefore, increasing complexity can reduce bias but may increase variance.
